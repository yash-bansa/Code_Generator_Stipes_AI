```python
import os
import sys
import asyncio
import asyncpg
from pathlib import Path
from dotenv import load_dotenv
```

```python
load_dotenv()
SCHEMA_NAME = "agentic_poc"
```

```python
conn = await asyncpg.connect(
host=os.getenv('PG_DB_HOST'),
port=os.getenv('PG_DB_PORT'),
user=os.getenv('PG_DB_USER'),
password=os.getenv('PG_DB_PASSWORD'),
database=os.getenv('PG_DB_NAME')
)

rows = await conn.fetch(f"SELECT file_path from {SCHEMA_NAME}.class_trail_embeddings")
await conn.close()
```

```python
print(len(rows))
```

```python
for col in rows:
    print(col)
```

```python
for col in rows[0].keys():
    print(f"{col} = {rows[0][col]}")
```

### Delete content from table

```python
conn = await asyncpg.connect(
    host=os.getenv('PG_DB_HOST'),
    port=os.getenv('PG_DB_PORT'),
    user=os.getenv('PG_DB_USER'),
    password=os.getenv('PG_DB_PASSWORD'),
    database=os.getenv('PG_DB_NAME')
)

# Delete all contents from the table
await conn.execute(f"DELETE FROM {SCHEMA_NAME}.class_trail_embeddings")

await conn.close()
```

### Get all the tables

```python
conn = await asyncpg.connect(
    host=os.getenv('PG_DB_HOST'),
    port=os.getenv('PG_DB_PORT'),
    user=os.getenv('PG_DB_USER'),
    password=os.getenv('PG_DB_PASSWORD'),
    database=os.getenv('PG_DB_NAME')
)

tables_query = f"""
            SELECT table_name
            FROM information_schema.tables
            WHERE table_schema = '{SCHEMA_NAME}'
            AND table_type = 'BASE TABLE';
        """
rows = await conn.fetch(tables_query)
await conn.close()
```

```python
for row in rows:
    print(row)
```
